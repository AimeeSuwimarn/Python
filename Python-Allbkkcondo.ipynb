```python
page = 1
name_list = []
Devname_list = []
price_list=[]
while page <= 102 :
    data = requests.get('https://www.checkraka.com/condo/?province=1&page=' + str(page))
    soup = bs4.BeautifulSoup(data.text)
    for c in soup.find_all('div',{'class':'product-highlight-card'}): 
        name_list.append(c.find('div',{'class':'card'}).find('a').text)
        Devname_list.append(c.find('div',{'class':'mid'}).text)
        price_list.append(c.find('div',{'class':'card'}).text)
    print('complete page number: ' ,page)
    page += 1
table = pd.DataFrame([name_list,Devname_list,price_list]).transpose()
table.columns = ['name','Devname','price']
table.set_index('name')
```

```python
table = table.set_index('name')
```

```python
table.to_excel('Allcondo.xlsx',engine='openpyxl')
```
